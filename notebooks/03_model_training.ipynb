## Import Libraries and Dependencies

```python
import pandas as pd
import numpy as np
import os
import joblib

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier

from xgboost import XGBClassifier
```

## Load Preprocessed Data

```python
import pandas as pd
import numpy as np

X_train = pd.read_csv("../Data/processed/X_train_ultra.csv")
X_test  = pd.read_csv("../Data/processed/X_test_ultra.csv")
y_train = pd.read_csv("../Data/processed/y_train_ultra.csv").values.ravel()
y_test  = pd.read_csv("../Data/processed/y_test_ultra.csv").values.ravel()

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("Target distribution:", np.bincount(y_train))
```

## Import  Classification Algorithms

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier,
    BaggingClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
```

## Define Model Zoo and Hyperparameters

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        min_samples_leaf=10,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        max_depth=8,
        random_state=42
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=500,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=300,
        learning_rate=0.05,
        random_state=42
    ),

    "Bagging": BaggingClassifier(
        n_estimators=300,
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=7,
        weights="distance"
    ),

    "Naive Bayes": GaussianNB()
}
```

## Train All Models

```python
trained_models = {}

for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)
    trained_models[name] = model

print("✅ All models trained")
```

## Evaluate and Compare Model Performance

```python
from sklearn.metrics import accuracy_score, f1_score

results = {}

for name, model in trained_models.items():
    y_pred = model.predict(X_test)
    results[name] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "F1-score": f1_score(y_test, y_pred)
    }

results_df = pd.DataFrame(results).T.sort_values("F1-score", ascending=False)
results_df
```

## Save Trained Models for Deployment

```python
os.makedirs("../models", exist_ok=True)

for name, model in trained_models.items():
    joblib.dump(
        model,
        f"../models/{name.replace(' ','_').lower()}_ultra.pkl"
    )

print("✅ All models saved successfully")
```
